```python
from PIL import Image

imagem = Image.open("image.jpg")

largura, altura = imagem.size

imagem_cinza = Image.new("L", imagem.size)
```

```python
for y in range(altura):
    for x in range(largura):
        R, G, B = imagem.getpixel((x, y))

        cinza = int(0.299 * R + 0.587 * G + 0.114 * B)

        imagem_cinza.putpixel((x, y), cinza)
```

```python
import matplotlib.pyplot as plt

plt.imshow(imagem_cinza, cmap="gray")
plt.axis("off")
plt.show()
```

```python
limiar = 128
imagem_binaria = Image.new("L", imagem.size)

for y in range(altura):
    for x in range(largura):

        cinza = imagem_cinza.getpixel((x, y))

        if cinza >= limiar:
            valor = 255
        else:
            valor = 0

        imagem_binaria.putpixel((x, y), valor)
```

```python
plt.imshow(imagem_binaria, cmap="gray")
plt.axis("off")
plt.show()
```
